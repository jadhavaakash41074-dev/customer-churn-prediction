```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore') # we are hiding the unnecessary warning 

# sets graph thems 
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (10 , 5)
plt.rcParams["axes.titlesize"] = 14
plt.rcParams["axes.labelsize"] = 12
```

```python
# Loading the data 
file_path  = r"A:\Downloads\data science projects\customer-churn-project\data\raw\telco_churn.csv"
df = pd.read_csv(file_path)
print("Data successfully loaded")
print(f"Total Rows : {df.shape[0]}")
print(f"Total Columns : {df.shape[1]}")
```

```python
display(df.head())

# to check the missing values
print("Data info ")
df.info()
```

```python
plt.Figure(figsize=(6 , 4))
ax = sns.countplot(data=df , x = "churn",palette= "Set2")

for p in ax.patches:
    ax.annotate(f"{p.get_height()}", (p.get_x() + p.get_width() / 2.,p.get_height()),
                ha = 'center', va = 'baseline', fontsize = 12, color = 'black', xytext=(0,5),
                 textcoords='offset points' )
plt.title("Churn Distribution (Target Variable)")
plt.xlabel("Churn (Yes/No)")
plt.ylabel("Number of Customers")
plt.show()

#percentage 
churn_rate = df["churn"].value_counts(normalize=True) * 100
print(f"Churn Percentage : \n{churn_rate.round(2)}")
```

```python
# to plots graph for EDA 
fig , axes = plt.subplots(1 , 2 , figsize = (14 , 5))

# 1 tenure(how many months customer can stayed) vs churn
sns.boxplot(data = df, x = "churn" , y="tenure_months", palette="Set2", ax=axes[0])
axes[0].set_title("Tenure vs Churn")


# monthly charges vs chrun 
sns.boxplot(data = df , x = "churn", y = "monthly_charges", palette="Set2" ,ax = axes[1])
axes[1].set_title("monthly_charges vs Churn")

plt.tight_layout()
plt.show()
```

```python
fig , axes = plt.subplots(1 , 2 , figsize=(16 , 5))

#1 . contract Type  vs Churn 

sns.countplot(data = df, x = "contract_type" , hue="churn", palette="Set2", ax = axes[0])
axes[0].set_title("Contract type vs Chrun ")

# 2 . Internet Service vs Chrun 

sns.countplot(data = df, x = "payment_method", hue = "churn", palette="Set2" , ax = axes[1])
axes[1].set_title("Payment Method vs Churn")
axes[1].tick_params(axis = 'x' , rotation = 45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8 , 5))
sns.countplot(data = df  , x = "customer_service_calls", hue = "churn", palette="Set2")
plt.title("Customer Service Calls vs Churn ")
plt.xlabel("Numbers of Customer service calls")
plt.ylabel("customer count")
plt.show()
```
